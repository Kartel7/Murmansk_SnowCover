```python
import pandas as pd
import geopandas as gpd
import numpy as np
from shapely import Point
```

Задаем имена метеостанций Мурманской области и их координаты, основываясь на открытых данных http://aisori-m.meteo.ru/

```python
data = {
    'Код': [22003, 22019, 22028, 22095, 22101, 22113, 22140, 22165, 22204, 22217, 22235, 22249, 22324, 22355, 22408],
    'Станция': ['Вайда-Губа', 'Полярное', 'Териберка', 'Колгуев Северный', 'Янискоски', 'Мурманск', 'Святой Нос', 'Канин Нос', 'Ковдор', 'Кандалакша', 'Краснощелье', 'Каневка', 'Умба', 'Сосновец остров', 'Калевала'],
    'Широта': [69.9333, 69.2, 69.2, 69.5333, 68.9667, 68.9667, 68.15, 68.65, 67.5667, 67.15, 67.35, 67.1333, 66.6833, 66.4833, 65.22],
    'Долгота': [31.9833, 33.4833, 35.1167, 49.0833, 28.7833, 33.05, 39.7667, 43.3, 30.45, 32.35, 37.05, 38.0667, 34.35, 40.6833, 31.22]
}
```

```python
# Создание DataFrame
df = pd.DataFrame(data)

# Создание столбца с геометрией
geometry = [Point(xy) for xy in zip(df['Долгота'], df['Широта'])]

# Создание GeoDataFrame
gdf = gpd.GeoDataFrame(df, geometry=geometry)
```

```python
# Создаем точки по координатам
[Point(xy) for xy in zip(df['Долгота'], df['Широта'])]
```

```python
# Задаем СК: WGS84
gdf.crs = 'EPSG:4326'
```

```python
gdf.crs
```

```python
gdf.columns = ['Code', 'Name', 'lat', 'lon', 'geometry']
```
